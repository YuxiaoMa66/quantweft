## Adding percentages loses compounding

Starting at 100, a 50% rise followed by a 50% fall ends at 75. The arithmetic mean is zero while the geometric mean is negative. Include initial wealth when computing drawdowns.

````python
returns = np.array([0.5, -0.5])
wealth = wealth_index(returns, initial_capital=100)
assert np.allclose(wealth, [100, 150, 75])
print({"wealth": wealth, "arithmetic_mean": returns.mean(),
       "geometric_mean": geometric_mean_return(returns),
       "drawdown": drawdown_path(wealth)})
````

## Same terminal wealth, different path risk

The two paths reorder the same three annual returns. Their CAGRs agree, while maximum losses from a prior peak are 52% and 40%.

````python
for sequence in ([0.5, -0.4, -0.2], [-0.4, 0.5, -0.2]):
    values = wealth_index(np.array(sequence), 100)
    print(values, cagr(100, float(values[-1]), 3), -drawdown_path(values).min())
````
